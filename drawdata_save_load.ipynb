# Как работает drawdata

**Для чего это вообще нужно?**

Очень удобный инструмент для Jupyter Notebook для создания тестовых данных кластеризации и классификации. Просто берём и рисуем

**Что здесь происходит?**

- создание датасета или редактирование
- загрузка из ранее сохранённого файла
- сохранение в файл
- график в Pandas

```python
from drawdata import ScatterWidget
import pandas as pd

from pathlib import Path
```

Создадим тестовый набор данных. Проверим, если сохранился сохранённый набор данных с прошлого запуска — загрузим его.

```python
csv_file = Path('df.csv')
dataset = ScatterWidget()
if csv_file.exists():
    dataset.data = pd.read_csv(csv_file).to_dict('records')
```

Отредактируем загруженное или нарисуем новое

```python
dataset
```

Сохраним изменения в файл

```python
dataset.data_as_pandas.to_csv(csv_file, index=False)
```

Загрузим данные из файла и нарисуем график средствами Pandas

```python
loaded = pd.read_csv(csv_file)
ax = loaded.plot.scatter(x='x', y='y', c='color', title='Загруженные данные')
```
