## Рукорисованный график

Можно рисовать такие графики, **matplotlib** поддерживает это из коробки. Нужно включить режим `xkcd`

```python
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np

# Включаем режим xkcd
plt.xkcd(scale=1, length=100, randomness=2)

# Явно задаем шрифт с поддержкой кириллицы
mpl.rcParams["font.family"] = "Excalifont"
mpl.rcParams["axes.unicode_minus"] = False  # не использовать типографский минус, использовать обычный, т.к. в подключаемом шрифте может не быть типографского

x = np.linspace(2018, 2026, 100)
y = np.exp((x - 2018) * 0.6) # Экспоненциальный рост

plt.figure(figsize=(10, 6))
plt.plot(x, y, color='red', linewidth=2)

plt.title('История звёздочек в рисованном стиле')
plt.xlabel('Дата')
plt.ylabel('GitHub звёзды')

plt.show()
```

## Параметры

`plt.xkcd(scale=1, length=100, randomness=2)`

- **scale** Сила искажения линии. меньше — линия ближе к обычной ровной; больше — сильнее “дрожит”.
- **length** Длина волны искажения. маленькое значение — частая мелкая дрожь; большое значение — более плавные длинные волны.
- **randomness** Случайность искажения. меньше — более регулярная, предсказуемая линия; больше — более хаотичная.

более аккуратно: `plt.xkcd(scale=0.5, length=100, randomness=1)`
сильнее “от руки”: `plt.xkcd(scale=2, length=80, randomness=4)`
